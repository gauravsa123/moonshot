```python
import pandas as pd
```

```python
ref_df = pd.read_csv("reference.csv")
ref_df['Mac'] = '/Users/gauravadke/Library/CloudStorage/OneDrive-MFPMichelin/work/'

ref_df['Relative_Path'] = (
    ref_df['Path']
    .str.replace(r'^.*?work\\', '', regex=True, case=False)
    .str.strip(' "')
)

ref_df['Relative_Path'] = ref_df['Relative_Path'].str.replace('\\', '/', regex=False)

ref_df.head()
```

```python
from pathlib import Path

wiki_root = Path("raw")

for topic, subtopic in ref_df[['Topic', 'Subtopic']].drop_duplicates().itertuples(index=False):
    # sanitize names to avoid invalid characters in folder names
    topic_dir = str(topic).strip().replace('/', '-')
    subtopic_dir = str(subtopic).strip().replace('/', '-')
    (wiki_root / topic_dir / subtopic_dir).mkdir(parents=True, exist_ok=True)

# display created structure
for p in sorted(wiki_root.rglob('*')):
    if p.is_dir():
        print(p)
```

```python
import shutil

copied, missing = 0, []

for row in ref_df.itertuples(index=False):
    if pd.isna(row.Relative_Path):
        continue

    src = Path(row.Mac) / row.Relative_Path
    dest_dir = wiki_root / str(row.Topic).strip().replace('/', '-') / str(row.Subtopic).strip().replace('/', '-')
    dest_dir.mkdir(parents=True, exist_ok=True)

    if src.is_file():
        shutil.copy2(src, dest_dir / src.name)
        copied += 1
        print(f"Copied: {src} to {dest_dir / src.name}")
    else:
        missing.append(str(src))
        print(f"Missing: {src}")

print(f"Copied: {copied}, Missing: {len(missing)}")
for m in missing:
    print(m)
```
